```python
import pandas as pd
import sqlalchemy
from sqlalchemy import create_engine
```

```python
df = pd.read_csv(
    "bank_analytics_clean.csv"
)
```

```python
engine = create_engine(
    "mysql+pymysql://root:arbaz%40123@localhost/insightbank"
)
```

```python
connection = engine.connect()

print("Database connected successfully")
```

```python
df.to_sql(
    name="bank_campaign_data",
    con=engine,
    if_exists="replace",
    index=False
)
```

```python
query = """

SELECT

    COUNT(*) AS total_customers,

    SUM(
        CASE
            WHEN y = 'yes'
            THEN 1
            ELSE 0
        END
    ) AS successful_conversions,

    ROUND(
        SUM(
            CASE
                WHEN y = 'yes'
                THEN 1
                ELSE 0
            END
        ) * 100.0 / COUNT(*),
        2
    ) AS conversion_rate

FROM bank_campaign_data;

"""
```

```python
kpi_results = pd.read_sql(
    query,
    con=engine
)

kpi_results
```
